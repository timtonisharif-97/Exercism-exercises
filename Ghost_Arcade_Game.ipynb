```python
def eat_ghost(power_pellet_active, touching_ghost):
  '''
    Determines whether player eats ghost

parameters:
    power_pellet_active(bool): does player have an active power pellet?
    touching_ghost(bool): is player touching the ghost?

returns:
    bool: does player eat ghost

  '''

  return power_pellet_active and touching_ghost


def score(Pac_Man_touching_pellet, Pac_Man_touching_dot):
  '''
        Check if Pac-Man Scored

  parameters:
        Pac_Man_touching_pellet(bool): is Pac-Man touching the power pellet?
        Pac_Man_touching_dot(bool): is Pac-Man touching the dot?

  returns:
        bool: Does Pac-Man score?
  '''

  return Pac_Man_touching_pellet or Pac_Man_touching_dot


def lose(Pac_Man_power_pellet_active, Pac_Man_touching_ghost):
  '''
      Shows if Pac-Man has lost

  parameters:
      Pac_Man_power_pellet_active(bool): does Pac-Man have power pellet active?
      Pac_Man_touching_ghost(bool): is Pac-Man touching ghost?

  returns:
      bool: Did Pac-Man lose?
  '''
  return not Pac_Man_power_pellet_active and Pac_Man_touching_ghost


def win(Pac_Man_eaten_all_dots, Pac_Man_power_pellet_active, Pac_Man_touching_ghost):
  '''
    Shows whether Pac-Man has won

 parameters:
     Pac_Man_eaten_all_dots(bool): did Pac-Man eat all dots?
     Pac_Man_power_pellet_active(bool); is Pac-Man power pellet active?
     Pac_Man_touching_ghost(bool): is Pac-Man touching ghost?

 returns:
     bool: does Pac-Man win?

  '''

  return Pac_Man_eaten_all_dots and not lose(Pac_Man_power_pellet_active, Pac_Man_touching_ghost)

```
