```python
def value_of_card(card):
  '''
      Returns the numerical value of the passed-in card

  parameter:
      card(str): passed-in card string

  returns:
      int: the numerical value of the passed-in card
  '''

  if card == "Ace":
    return 1
  if card in {"J", "Q", "K"}:
    return 10

  return int(card)


def higher_card(card_one, card_two):
  '''
      Shows which of the two cards has a higher scoring value

  paramters:
      card_one(str): first card received
      card_two(str); second card received

  returns:
      str: the card with the higher scoring value
  '''

  if value_of_card(card_one) > value_of_card(card_two):
    return card_one
  elif value_of_card(card_two) > value_of_card(card_one):
    return card_two

  return (card_one, card_two)




def value_of_ace(card_one, card_two):
  '''
      Determines if the upcoming ace will get a value of 1 or a value of 11

parameters:
      card_one(str): first card received
      card_two(str); second card received

returns:
      int: value of upcoming ace
  '''

  if "A" in (card_one, card_two):
    return 1
  if value_of_card(card_one) + value_of_card(card_two) > 10:
    return 1

  return 11


def is_blackjack(card_one, card_two):
  '''
      Determines if the cards is a blackjack

  parameters:
      card_one(str): first card received
      card_two(str); second card received

  returns:
      bool: whether two cards is a blackjack
  '''
  blackjack = card_one == "A" and card_two in {"10", "J", "Q", "K"} \
              or card_two == "A" and card_one in {"10", "J", "Q", "K"}


  return blackjack



def can_split_pairs(card_one, card_two):
  '''
      Determines if two-card hand can be split into two pairs

parameters:
      card_one(str): first card received
      card_two(str); second card received

returns:
      bool: whether two-card hand can be split into two pairs
  '''
  split = value_of_card(card_one) == value_of_card(card_two)

  return split

  def can_double_down(card_one, card_two):
    '''
      Determines if two-card hand can be "doubled down"

    parameters:
      card_one(str): first card received
      card_two(str); second card received

    returns:
      bool: whether two-card hand can be "doubled down"
    '''

  value_of_cards_dealt = value_of_card(card_one) + value_of_card(card_two)

  return value_of_cards_dealt in {9, 10, 11}





```
