```python
def is_criticality_balanced(temperature, neutrons_emitted):
  '''
      Checks reactor's criticality condition

  parameters:
      temperature(float): the temperature of the reactor
      neutrons_emitted(int): the number of neutrons emitted in the reactor

  returns:
      bool: reactor's criticality condition
  '''

  return temperature < 800 and neutrons_emitted > 500 and (temperature * neutrons_emitted < 500000)



def reactor_efficiency(voltage, current, theoretical_max_power):
  '''
      Shows the efficiency band of the nuclear reactor

  parameters:
      voltage(float): the amount of volts in the reactor
      current(float): the amount of current in the reactor
      theoretical_max_power(float): the max power reactor can handle in theory

  returns:
      str: the efficiency band of the reactor
  '''
  generated_power =  voltage * current
  efficiency = (generated_power/theoretical_max_power)*100

  if efficiency >= 80:
    return "green"
  if 60 <= efficiency < 80:
    return "orange"
  if 30 <= efficiency < 60:
    return "red"

  return "black"


def fail_safe(temperature, neutrons_produced_per_second, threshold):
  '''
      Show status of reactor's criticality threshold

  parameters:
      temperature(float): the temperature of the reactor
      neutrons_produced_per_second(int): neutrons produced in reactor per second
      threshold(float): reactor's criticality level

 returns:
      str: status of reactor's criticality threshold
  '''
  test = temperature * neutrons_produced_per_second

  if test < (0.9 * threshold):
    return "LOW"
  if (0.9 * threshold) <= test <= (1.1 * threshold):
    return "NORMAL"

  return "DANGER"





```
