```python
def capitalize_title(paper_title):
  '''
      Capitalizes First Letter Of Each Word Of A Title

parameters:
      paper_title(str): title of paper

returns:
      str: title case of a title
  '''
  formatted_title = paper_title.title()

  return formatted_title


def check_sentence_ending(sentence):
  '''
     checks if sentence ends with a period

parameters:
      sentences(str): any sentence in paper

returns:
      bool: if sentence ends with a period
  '''
  end_of_sentence = sentence.endswith(".")

  return end_of_sentence



def clean_up_spacing(sentence):
  '''
     cleans up spacing in sentence

parameters:
      sentences(str): any sentence in paper

returns:
      str: removed extra whitespace in sentences
    '''
  formatted_sentence = sentence.strip()

  return formatted_sentence


def replace_word_choice(sentence, old_word, new_word):
  '''
      replaces words with synonyms

parameters:
      sentences(str): any sentence in paper
      old_word(str): old adjective used in paper
      new_word(str): adjective used to replace old adjective

returns:
      str: synonyms to old words
    '''
  new_sentence = sentence.replace(old_word, new_word)

  return   new_sentence



```
