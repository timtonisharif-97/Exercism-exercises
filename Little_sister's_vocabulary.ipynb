```python

def add_prefix_un(word):
  '''
      Returns word with a new un prefixed word

  parameters:
      word(str): sister's chosen word

 returns:
      str: un prefixed word
  '''
  return "un" + word

def make_word_groups(vocab_words):
  '''
      Returns vocabulary words with selected prefixes

  parameters:
      vocab_word(str): sister's chosen list of words

  returns:
      str:
    '''
  prefix = vocab_words[0]
  words = vocab_words[1:]
  separator = " :: " + prefix
  prefix_words = separator.join(words)

  return prefix + separator + prefix_words


def remove_suffix_ness(word):
  '''
      Removes suffix ness from words

  parameters:
      word(str): sister's chosen words

 returns:
      str: word without suffix ness madness
  '''
  stem_word = word[0:-4]

  if stem_word.endswith("i"):
    return stem_word[0:-1] + "y"

  return stem_word



def adjective_to_verb(sentence, index):
  '''
      Returns extracted adjective as a verb

  parameters:
      sentence(str): sister's chosen vocabulary words
      index(int): position of the chosen word

 returns:
      str: extracted adjective as a verb
  '''
  words = sentence.split(" ")
  adjective = words[index]
  adjective = adjective.strip(".")

  return adjective + "en"






```
