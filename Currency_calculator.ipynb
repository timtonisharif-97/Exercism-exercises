```python
def exchange_money(budget, exchange_rate):
  '''
      Calculates currency exchange

  parameters:
      budget(float): Amount of money to be exchanged
      exchange_rate(float): rate at which a local currency is exchange for a foreign

  returns:
      float: the equivalent amount of a foreign currency in a local one
  '''

  return budget / exchange_rate


def get_change( budget, exchanging_value):
  '''
      Calculates the remaining change

  paramters:
      budget(float): amount of money to be exchanged
      exchanging_value(float): amount of money exchanged

  returns:
      int: the remaining change after exchange
  '''

  return budget - exchanging_value


def get_value_of_bills(denomination, number_of_bills):
  '''
      Calculates the total value of the exchanged bills

  parameters:
      denomination(int): The value of a single bill.
      number_of_bills(int): The total number of bills

  returns:
      int: the total value of the exchanged bills
  '''
  return denomination * number_of_bills


def get_number_of_bills(amount, denomination):
  '''
    Calculates the number of exchanged bills received in a given denomination

  paramters:
    amount(float): total amount of exchanged currency
    denomination(int): single unit value of exchanged currency

  returns:
    int: the number of exchanged bills received in a given denomination

  '''
  return amount // denomination


def get_leftover_of_bills(amount, denomination):
  '''
    Calculates the leftover exchanged currency bills

  parameters:
    amount(float): total amount of exchanged currency
    denomination(int): single unit value of exchanged currency

  returns:
    int: the leftover exchanged currency bills
  '''
  return amount % denomination


def exchangeable_value(budget, exchange_rate, spread, denomination):
  '''
      calculates the maximum value of the exchanged currency after the exchange rate plus the spread

  parameters:
      budget(float): total amount of currency available for exchnage
      exchange_rate(float): rate at which a local currency is exchange for a foreign
      spread(float): exchange fee
      denomination(int): single unit value of exchanged currency

  returns:
      int: the maximum value of the exchanged currency after the exchange rate plus the spread
  '''
  spread = spread / 100
  new_exchange_rate = exchange_rate + (spread * exchange_rate)
  exchange_amount = budget / new_exchange_rate

  return get_number_of_bills(exchange_amount, denomination) * denomination
```
